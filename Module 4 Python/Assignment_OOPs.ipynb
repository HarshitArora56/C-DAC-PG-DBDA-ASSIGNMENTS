```python
# 8.	Create a class Teacher with name, age, and salary attributes, 
# where salary must be a private attribute that cannot be accessed outside the class.

class Teacher:
    def __init__(self, name, age, __salary):
        self._name = name
        self.age = age
        self.__salary = __salary
    def __str__(self):
       return f'Name: {self._name:<20} Age: {self.age:<10} Salary: {self.__salary:<10}'

        

t = Teacher("ABCD", '35', 40000)
t.__salary = 50000 # salary not accessible outside class hence no change in output
print(t)
```
