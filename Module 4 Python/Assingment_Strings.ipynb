```python
# Write a function that takes a string as input and returns a list of all the words in the string that are longer than a specified length. 
# The function should take two arguments: the string and the minimum length of a word. 
# For example, if the input string is "The quick brown fox jumps over the lazy dog" and the minimum length is 4, 
# the function should return ["quick", "brown", "jumps"].

def words_longer_than(s, num):
    str_lst = s.split(' ')
    res = []
    for i in str_lst:
        if len(i) > num:
            res.append(i)
    return res

s = "The quick brown fox jumps over the lazy dog"
words_longer_than(s, 4)
```

```python
# Write a function that takes a string as input and returns the string with all the vowels converted to uppercase. 
# For example, if the input string is "hello world", the function should return "hEllO wOrld".

def vowels_uppercase(s):
    res = ''
    for i in s:
        if i not in 'aeiou':
            res += i
        else:
            res += i.upper()
    return res

    
s = input('Enter the String: ')
print(vowels_uppercase(s))
```

```python
# Write a program that reads in a list of words from the user, separated by spaces, and then prints out the length of the longest word. 
# For example, if the user enters "the quick brown fox jumps over the lazy dog", the program should print out "5".

def longest_word_length(s):
    max_word = 0
    for word in s:
        if len(word) > max_word:
            max_word = len(word)

    return max_word

s = input("Enter the sentence: ").split()
print(f'Longest word length: {longest_word_length(s)}')
            
```

```python
# Write a Python function that takes two strings and concatenates them, 
# but only if the second string starts with the first three letters of the first string. 
# Otherwise, the function should return the original first string.

def conditional_concat(s1, s2):
    if s2.startswith(s1[:3]):
        return s1 + " " + s2
    else:
        return s1

print(conditional_concat('hello', 'helicopter'))
print(conditional_concat('hello', 'world'))
```

```python
def most_freq_char(s):
    freq = {}
    for c in s:
        if c in freq:
            freq[c] = freq[c] + 1
        else:
            freq[c] = 1

    res = ''
    sort_sd = sorted(freq, key = lambda char : freq[char], reverse = True)

    for c in sort_sd:
        res = res + c * freq[c]
        
    
    return res

s = 'banana'
print(most_freq_char(s))


    
```

```python
l = [1,2,3,4,5,6,7,8,9]

target = 12

def check_target(l, target):
    for i in range(len(l)):
        curr_sum = 0
        curr_sum += l[i]
        for j in range(i+1,len(l)):
            curr_sum += l[j]
            if curr_sum == target:
                return [i,j]
            elif curr_sum > target:
                break
            
        
    
print(check_target(l,target))
```

```python
lst = ['eat','tea','tan','ate', 'nat', 'bat']

def anagram(s1, s2):
    return sorted(s1) == sorted(s2)



def group_anagram(lst):
    anagram_dict = {}
    
    for word in lst:
        sorted_word = "".join(sorted(word))
        
        if sorted_word not in anagram_dict:
            anagram_dict[sorted_word] = []
        anagram_dict[sorted_word].append(word)
        
    return list(anagram_dict.values())

group_anagram(lst)
```

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None

class Linked_list:
    def __init__(self):
        self.root = None


    def insert_right(self, n):
        if self.root == None:
            self.root = n
        else:
            t = self.root
            while t.next != None:
                t = t.next
            t.next = n


    def find_in_list(self, key):
        t = self.root
        count = 1
        while t.data != key:
            t = t.next
            count += 1
        return count
        

    def print_list(self):
        if self.root == None:
            print('Empty List')
        else:
            t = self.root
            while t != None:
                print(f'|{t.data}|', end = '->')
                t = t.next
                
        
        
```

```python
l = Linked_list()

l.insert_right(Node(10))
l.insert_right(Node(5))
l.insert_right(Node(7))
l.insert_right(Node(20))
l.insert_right(Node(30))
l.insert_right(Node(100))
l.insert_right(Node(200))
print(l.find_in_list(20))
l.print_list()
```

```python
class stack:
    def __init__(self):
        self.stack_l = []


    def push(self, data): # entering the data
        self.stack_l.append(data)

    def pop(self): # removes the top most / last entered data
        return self.stack_l.pop()

    def peek(self): # view the last entered data / returns last entered data
        return self.stack_l[-1]

    def print_stack(self):
        for i in reversed(self.stack_l):
            print(i)

    def is_empty(self):
        return len(self.stack_l) == 0


    
```

```python
s = stack()
s.push(10)
s.push(20)
s.push(30)
s.push(40)
s.push(50)
s.push(60)
s.pop()
s.peek()
s.print_stack()
s.is_empty()
```

```python
class Queue:
    def __init__(self):
        self.queue_l = []

    def enqueue(self, data):
        self.queue_l.append(data)

    def dequeue(self):
        return self.queue_l.pop(0)

    def print_queue(self):
        for i in self.queue_l:
            print(i)

    def is_empty(self):
        return len(self.queue_l) == 0
```

```python
q = Queue()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.enqueue(40)
q.enqueue(50)

q.dequeue()
q.print_queue()
```

```python
brackets = input('Enter the brakets: ')

def check_brackets(b):
    s = stack()
    for i in b:
        if i in '({[':
            s.push(i)
        elif i == ')':
            if s.is_empty() or s.pop() != '(':
                return False
        elif i == '}':
            if s.is_empty() or s.pop() != '{':
                return False
        elif i == ']':
            if s.is_empty() or s.pop() != '[':
                return False
    return s.is_empty()
        
print(check_brackets(brackets))
```

```python
def c_b(b):
    map_b = {
        ')' : '(',
        '}' : '{',
        ']' : '['
    }

    l = []
    for key in b:
        if key in map_b.values():
            l.append(key)
        elif key in map_b:
            if not l or l[-1] != map_b[key]:
                return False
            l.pop()
    return not l
c_b(brackets)
```

```python
my_dict = {}

print("Dictionary Builder")
print("=" * 40)
print("Enter key-value pairs to build a dictionary.")
print("Type 'done' when finished.\n")

while True:
    key = input("Enter key (or 'done' to finish): ").strip()
    
    if key.lower() == 'done':
        break
    
    if not key:
        print("Key cannot be empty. Try again.\n")
        continue
    
    value = input(f"Enter value for '{key}': ").strip()
    
    if not value:
        print("Value cannot be empty. Try again.\n")
        continue
    
    my_dict[key] = value
    print(f"Added: {key} → {value}\n")

print("\n" + "=" * 40)
print("Final Dictionary:")
print(my_dict)
```
