```python
# Ques 1
num = int(input("Enter the number: "))
if num % 2 == 0:
    print(f'{num} is even')
else:
    print(f'{num} is odd')
```

```python
# Ques 2
year = int(input("Enter the year: "))

if year % 100 == 0:
    if year % 400 == 0:
        print(f'{year} This is a centenary year and leap year')
else:
    if year % 4 == 0:
        print(f'{year} is a leap year')
```

```python
# Ques 3

pallindrome_check = str(input("Enter the String to check pallindrome: "))

rev_str = pallindrome_check[::-1]

if pallindrome_check == rev_str:
    print(f'{pallindrome_check} is a pallindrome')
else:
    print('')
```
