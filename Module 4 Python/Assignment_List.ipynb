```python
# Write a Python program to remove duplicates from a list.
def remove_duplicates(input_list):
    seen = set()
    res = []
    for i in input_list:
        if i not in seen:
            seen.add(i)
            res.append(i)
    return res

list1 = [1,2,3,1,2,4,5,3,6,8,3,6]
final_list = remove_duplicates(list1)
print(f'Original List: {list1}')
print(f'List without duplicates: {final_list}')
```

```python
# Write a Python program to get the sum of all the elements in a list.
def sum_of_elements(input_list):
    total = 0
    for i in input_list:
        total += i
    return total

list1 = [1,2,3,1,2,4,5,3,6,8,3,6]

print(f'Total of all elements: {sum_of_elements(list1)}')
```

```python
# Write a Python program to find the largest number in a list.
def largest_num(input_list):
    high = input_list[0]
    for i in input_list:
        if i > high:
            high = i
    return high

list3 = [2,5,6,20,11,7,1]
print(f'Highest Element: {largest_num(list3)}')    
```

```python
# Write a Python program to sort a list in descending order.
def sort_desc(input_lst):
    for i in range(len(input_lst)):
        for j in range(i+1,len(input_lst)):
            if input_lst[i] < input_lst[j]:
                input_lst[i], input_lst[j] = input_lst[j],input_lst[i]
    return input_lst

sort_desc(lst)
```

```python
# Write a Python program to reverse a list.
def rev_lst(input_lst):
    rev = []
    for i in range(len(input_lst)-1,-1,-1):
        rev.append(input_lst[i])
    return rev
lst = [9,7,5,3,1]
rev_lst(lst)
```

```python
# Write a Python program to find the common elements between two lists.
def common_element(lst1, lst2):
    ce = set()
    for i in lst1:
        for j in lst2:
            if i == j:
                ce.add(i)

    return ce

l1 = [9,7,5,3,1]
l2 = [1,2,3,4,5,6,7,8,9]
common_element(l2,l1)
```

```python
# Write a Python function that takes a list of numbers as input and returns the sum of all the even numbers in the list. 
# You can assume that the input list only contains integers.
def sum_even(lst : list[int]):
    se = 0
    for i in lst:
        if i % 2 == 0:
            se += i
    return se

l = [1,20,48,2,3,24,5,7,99,6]
sum_even(l)
```

```python
# Write a Python function that takes two lists as input and returns a new list that contains all the elements that are common between the two input lists. 
# You can assume that the input lists only contain integers, and that the output list should not contain any duplicate elements.
def common(lst1, lst2):
    ce = []
    for i in lst1:
        for j in lst2:
            if i == j:
                if i not in ce:
                    ce.append(i)
    return ce

l1 = [9,7,5,3,1]
l2 = [1,2,3,4,5,6,7,8,9]
common(l2,l1)
```
