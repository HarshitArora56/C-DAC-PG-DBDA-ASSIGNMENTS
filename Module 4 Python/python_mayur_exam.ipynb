```python
class Book:
    def __init__(self,title,author,publication_year,genre,num_pages):
        self.title=title
        self.author=author
        self.publication_year=publication_year
        self.numpages=numpages

    def read(self,num_pages):
        self.num_pages=num_pages
        self.num_pages=self.numpages-self.num_pages
        print(f'num of pages is {self.num_pages}')

    def change_genre(self,new_genre):
        self.new_genre=new_genre
        self.genre=new_genre
        print(f'New genre is {self.genre}')

    def set_current_year(self,current_year):
        self.current_year=current_year

    def get_age(self):
        age=self.current_year-self.publication_year
        return f'Age is {age}'

    
        
   
```

```python
title=input('Enter book name: ')
author=input('Enter author name: ')
publication_year=int(input('Enter publication year: '))
genre=input("Enter genre")
numpages=int(input('Enter number of pages'))

b=Book(title,author,publication_year,genre,numpages)
```

```python
def anagram(s1,s2):
    cnt=0
    cnt1=0
    for i in s1:
        cnt+=1
    for i in s2:
        cnt1+=1
        
    for i in s1:
        if i not in s2:
            return False
        elif i in s2 and cnt1==cnt:
            return True
        else:
            return False
```
