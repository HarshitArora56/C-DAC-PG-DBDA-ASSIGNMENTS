```python
# 3. Write a Pandas program to add, subtract, multiple and divide two Pandas Series. 
# Sample Series: [2, 4, 6, 8, 10], [1, 3, 5, 7, 9]
sr1 = pd.Series([2, 4, 6, 8, 10])
sr2 = pd.Series([1, 3, 5, 7, 9])

add_series = sr1 + sr2
sub_series = sr1 - sr2
mul_series = sr1 * sr2
div_series = sr1 / sr2
print(f'add - \n{add_series}\n\nsub - \n{sub_series}\n\nmultiply - \n{mul_series}\n\ndivide - \n{div_series}')
```

```python
# 10. Write a Pandas program to convert a Series of lists to one Series. 
l1 = ['Red', 'Green', 'White']
l2 = ['Red', 'Black']
l3 = ['Yellow']
sol = pd.Series([l1,l2,l3])
```

```python
# method 1
res_s = sol.explode()
print(res_s.reset_index(drop=True))
```

```python
# method 2
res_s = pd.concat(pd.Series(x) for x in sol)
print(res_s.reset_index(drop=True))
```

```python
# 15. Write a Pandas program to create the mean and standard deviation of the data of a given Series. 
s1 = pd.Series([1,2,3,4,5,6,7,8,9,5,3])

mean_s1 = s1.mean()
print(f'Mean: {mean_s1}')
std_dev = s1.std()
print(f'Standart Deviation: {std_dev}')
```

```python
# 16. Write a Pandas program to get the items of a given series not present in another given series. 
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([2,4,6,8,10])

res_s = s1.iloc[np.where(~s1.isin(s2))]
print(f'Result: \n{res_s}')
```

```python
# 17. Write a Pandas program to get the items which are not common of two given series. 
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([2,4,6,8,10])

rs1 = s1.iloc[np.where(~s1.isin(s2))]
rs2 = s2.iloc[np.where(~s2.isin(s1))]

res_s = pd.concat([rs1,rs2])

print(f'Result: \n{res_s}')
```
